```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
import os
os.chdir("/Users/Isa García/Downloads/1/Storytelling y casos de éxito")
df = pd.read_csv("Productos.csv")
df
```

```python
# Identificar los valores nulos por columna
print("Valores nulos antes de limpiar:")
print(df.isnull().sum())
```

```python
# Eliminar filas que tengan valores nulos
df = df.dropna()
```

```python
# Verificar que ya no existan valores nulos
print("Valores nulos después de limpiar:")
print(df.isnull().sum())
```

```python
# Mostrar los primeros registros 
df.head()
```

```python
#Identificar todas las marcas
marcas = df["BRAND"].dropna().unique()
# Mostrar las marcas disponibles
print("Marcas disponibles en la base de datos:")
for marca in sorted(marcas):
    print(marca)
# Contar cuántas marcas existen
print("Total de marcas:", len(marcas))
```

```python
# Definir las marcas clave
marcas_clave = ["VANISH", "LYSOL"]
# Filtrar el DataFrame únicamente con las marcas seleccionadas
df_filtrado = df[df["BRAND"].isin(marcas_clave)].copy()
# Mostrar los primeros registros
print("Primeros registros de las marcas seleccionadas:")
display(df_filtrado.head())
```

```python
# Verificar las marcas incluidas
print("Marcas en df_filtrado:", df_filtrado["BRAND"].unique())
# Contar registros por marca
print(df_filtrado["BRAND"].value_counts())
# Total de registros filtrados
print("Total de registros:", len(df_filtrado))
```

```python
# Convertir DATE a formato fecha
df_filtrado["DATE"] = pd.to_datetime(df_filtrado["DATE"])
# Agrupar ventas por marca y mes
ventas_mensuales = (df_filtrado.groupby(["BRAND", pd.Grouper(key="DATE", freq="MS")])["TOTAL_UNIT_SALES"].sum().reset_index())
# Ordenar por marca y fecha
ventas_mensuales = ventas_mensuales.sort_values(["BRAND", "DATE"])
# Mostrar resultados
display(ventas_mensuales.head())
```

```python
#Divide los datos en conjuntos de entrenamiento (training) y prueba (testing) para validar el modelo.
# Separar ventas por marca
df_vanish = ventas_mensuales[ventas_mensuales["BRAND"] == "VANISH"].sort_values("DATE").copy()
df_lysol = ventas_mensuales[ventas_mensuales["BRAND"] == "LYSOL"].sort_values("DATE").copy()
#Datos por venta mensual dividida.
print("Datos por venta mensual dividida para VANISH:", len(df_vanish))
print("Datos por venta mensual dividida para LYSOL:", len(df_lysol))
```

```python
#Datos de entrenamiento de VANISH al 90% de entrenamiento y 10% de pruebas
n_vanish = int(len(df_vanish) * 0.90)
train_vanish = df_vanish["TOTAL_UNIT_SALES"].iloc[:n_vanish]
test_vanish = df_vanish["TOTAL_UNIT_SALES"].iloc[n_vanish:]

print("VANISH")
print("Datos de entrenamiento:", len(train_vanish))
print("Datos de prueba:", len(test_vanish))
```

```python
#Datos de entrenamiento de LYSOL al 90% de entrenamiento y 10% de pruebas
n_lysol = int(len(df_lysol) * 0.90)
train_lysol = df_lysol["TOTAL_UNIT_SALES"].iloc[:n_lysol]
test_lysol = df_lysol["TOTAL_UNIT_SALES"].iloc[n_lysol:]
print("LYSOL")
print("Datos de entrenamiento:", len(train_lysol))
print("Datos de prueba:", len(test_lysol))
```

### Modelo de Regresión Simple para pronosticar las ventas de dos marcas: VANISH y LYSOL

```python
# Variable independiente 
X_train_vanish = np.arange(len(train_vanish)).reshape(-1, 1)
X_test_vanish = np.arange(len(train_vanish), len(train_vanish) + len(test_vanish)).reshape(-1, 1)

X_train_lysol = np.arange(len(train_lysol)).reshape(-1, 1)
X_test_lysol = np.arange(len(train_lysol),len(train_lysol) + len(test_lysol)).reshape(-1, 1)
```

```python
from sklearn.linear_model import LinearRegression
# Crear modelos
modelo_vanish = LinearRegression()
modelo_lysol = LinearRegression()
# Modelo de  VANISH
modelo_vanish.fit(X_train_vanish,train_vanish)
# Modelo de  LYSOL
modelo_lysol.fit(X_train_lysol,train_lysol)

print("Modelos regresión líneal")

print("VANISH")
print("Intercepto:", modelo_vanish.intercept_)
print("Pendiente:", modelo_vanish.coef_[0])

print("LYSOL")
print("Intercepto:", modelo_lysol.intercept_)
print("Pendiente:", modelo_lysol.coef_[0])
```

```python
#Prediccion sobre TESTING
# Predicciones VANISH
pred_vanish = modelo_vanish.predict(X_test_vanish)
# Predicciones LYSOL
pred_lysol = modelo_lysol.predict(X_test_lysol)

print("Predicciones VANISH:")
display(pd.DataFrame({"Fecha": df_vanish["DATE"].iloc[n_vanish:].values, "Ventas_reales": test_vanish.values,
                      "Ventas_predichas": pred_vanish}))
print("Predicciones LYSOL:")
display(pd.DataFrame({"Fecha": df_lysol["DATE"].iloc[n_lysol:].values,"Ventas_reales": test_lysol.values,
                      "Ventas_predichas": pred_lysol}))
```

```python
#Evaluar el modelo
from sklearn.metrics import mean_absolute_error, mean_squared_error
# Evaluación VANISH
mae_vanish = mean_absolute_error(test_vanish, pred_vanish)
rmse_vanish = np.sqrt(mean_squared_error(test_vanish, pred_vanish))
# Evaluación LYSOL
mae_lysol = mean_absolute_error(test_lysol, pred_lysol)
rmse_lysol = np.sqrt(mean_squared_error(test_lysol, pred_lysol))
# Resultados VANISH
print("===== RESULTADOS VANISH =====")
print("MAE:", mae_vanish)
print("RMSE:", rmse_vanish)
# Resultados LYSOL
print("\n===== RESULTADOS LYSOL =====")
print("MAE:", mae_lysol)
print("RMSE:", rmse_lysol)
```

```python
# Error relativo VANISH
error_relativo_vanish = (mae_vanish / test_vanish.mean()) * 100

# Error relativo LYSOL
error_relativo_lysol = (mae_lysol / test_lysol.mean()) * 100

print("VANISH:", error_relativo_vanish, "%")
print("LYSOL:", error_relativo_lysol, "%")
```

```python
def interpretar_error(error):
    if error < 5:
        return "Muy bueno"
    elif error < 10:
        return "Bueno"
    elif error < 20:
        return "Moderado"
    elif error <= 30:
        return "Alto"
    else:
        return "Muy alto"
# Interpretación VANISH
print("VANISH:", interpretar_error(error_relativo_vanish))
# Interpretación LYSOL
print("LYSOL:", interpretar_error(error_relativo_lysol))
```

```python
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
# Modelo ARIMA para VANISH

# ARIMA(0,1,0)
mod_ar1_vanish = ARIMA(train_vanish, order=(0,1,0))
res_ar1_vanish = mod_ar1_vanish.fit()
print("1. AIC ARIMA(0,1,0) VANISH:", res_ar1_vanish.aic)

# ARIMA(1,1,0)
mod_ar2_vanish = ARIMA(train_vanish, order=(1,1,0))
res_ar2_vanish = mod_ar2_vanish.fit()
print("2. AIC ARIMA(1,1,0) VANISH:", res_ar2_vanish.aic)

# ARIMA(0,1,1)
mod_ar3_vanish = ARIMA(train_vanish, order=(0,1,1))
res_ar3_vanish = mod_ar3_vanish.fit()
print("3. AIC ARIMA(0,1,1) VANISH:", res_ar3_vanish.aic)

# ARIMA(1,1,1)
mod_ar4_vanish = ARIMA(train_vanish, order=(1,1,1))
res_ar4_vanish = mod_ar4_vanish.fit()
print("4. AIC ARIMA(1,1,1) VANISH:", res_ar4_vanish.aic)

# ARIMA(2,1,0)
mod_ar5_vanish = ARIMA(train_vanish, order=(2,1,0))
res_ar5_vanish = mod_ar5_vanish.fit()
print("5. AIC ARIMA(2,1,0) VANISH:", res_ar5_vanish.aic)

# ARIMA(0,1,2)
mod_ar6_vanish = ARIMA(train_vanish, order=(0,1,2))
res_ar6_vanish = mod_ar6_vanish.fit()
print("6. AIC ARIMA(0,1,2) VANISH:", res_ar6_vanish.aic)
```

```python
# Comparar los AIC para VANISH
aic_vanish = [ res_ar1_vanish.aic, res_ar2_vanish.aic, res_ar3_vanish.aic, res_ar4_vanish.aic, res_ar5_vanish.aic, res_ar6_vanish.aic]
modelos_vanish = [ res_ar1_vanish, res_ar2_vanish, res_ar3_vanish, res_ar4_vanish, res_ar5_vanish,res_ar6_vanish]
ordenes = [ (0,1,0), (1,1,0), (0,1,1), (1,1,1), (2,1,0), (0,1,2)]

posicion_vanish = np.argmin(aic_vanish)

mejor_modelo_vanish = modelos_vanish[posicion_vanish]
mejor_orden_vanish = ordenes[posicion_vanish]

print("MEJOR MODELO VANISH")
print("Mejor modelo ARIMA:", mejor_orden_vanish)
print("AIC mínimo:", mejor_modelo_vanish.aic)
```

```python
# Modelo ARIMA para LYSOL

# ARIMA(0,1,0)
mod_ar1_lysol = ARIMA(train_lysol, order=(0,1,0))
res_ar1_lysol = mod_ar1_lysol.fit()
print("1. AIC ARIMA(0,1,0) LYSOL:", res_ar1_lysol.aic)

# ARIMA(1,1,0)
mod_ar2_lysol = ARIMA(train_lysol, order=(1,1,0))
res_ar2_lysol = mod_ar2_lysol.fit()
print("2. AIC ARIMA(1,1,0) LYSOL:", res_ar2_lysol.aic)

# ARIMA(0,1,1)
mod_ar3_lysol = ARIMA(train_lysol, order=(0,1,1))
res_ar3_lysol = mod_ar3_lysol.fit()
print("3. AIC ARIMA(0,1,1) LYSOL:", res_ar3_lysol.aic)

# ARIMA(1,1,1)
mod_ar4_lysol = ARIMA(train_lysol, order=(1,1,1))
res_ar4_lysol = mod_ar4_lysol.fit()
print("4. AIC ARIMA(1,1,1) LYSOL:", res_ar4_lysol.aic)

# ARIMA(2,1,0)
mod_ar5_lysol = ARIMA(train_lysol, order=(2,1,0))
res_ar5_lysol = mod_ar5_lysol.fit()
print("5. AIC ARIMA(2,1,0) LYSOL:", res_ar5_lysol.aic)

# ARIMA(0,1,2)
mod_ar6_lysol = ARIMA(train_lysol, order=(0,1,2))
res_ar6_lysol = mod_ar6_lysol.fit()
print("6. AIC ARIMA(0,1,2) LYSOL:", res_ar6_lysol.aic)
```

```python
# Comparar los AIC para LYSOL 
aic_lysol = [ res_ar1_lysol.aic, res_ar2_lysol.aic, res_ar3_lysol.aic, res_ar4_lysol.aic, res_ar5_lysol.aic, res_ar6_lysol.aic]
modelos_lysol = [ res_ar1_lysol, res_ar2_lysol, res_ar3_lysol, res_ar4_lysol,  res_ar5_lysol,  res_ar6_lysol]
posicion_lysol = np.argmin(aic_lysol)

mejor_modelo_lysol = modelos_lysol[posicion_lysol]
mejor_orden_lysol = ordenes[posicion_lysol]

print("MEJOR MODELO LYSOL")
print("Mejor modelo ARIMA:", mejor_orden_lysol)
print("AIC mínimo:", mejor_modelo_lysol.aic)
```

```python
# Pronóstico para los datos de prueba
predicciones_vanish_test = mejor_modelo_vanish.forecast(steps=len(test_vanish))

# Intervalos de confianza del 90%
intervalos_vanish = mejor_modelo_vanish.get_forecast(steps=len(test_vanish)).conf_int(alpha=0.10)

print("RESULTADOS ARIMA - VANISH")
print("VENTAS REALES:")
print(test_vanish.values)
print("PREDICCIONES:")
print(predicciones_vanish_test.values)
print("LÍMITE INFERIOR (90%):")
print(intervalos_vanish.iloc[:, 0].values)
print("LÍMITE SUPERIOR (90%):")
print(intervalos_vanish.iloc[:, 1].values)
```

```python
# Pronóstico para los datos de prueba
predicciones_lysol_test = mejor_modelo_lysol.forecast( steps=len(test_lysol))

# Intervalos de confianza del 90%
intervalos_lysol = mejor_modelo_lysol.get_forecast( steps=len(test_lysol)).conf_int(alpha=0.10)

print("RESULTADOS ARIMA - LYSOL")
print("VENTAS REALES:")
print(test_lysol.values)
print("PREDICCIONES:")
print(predicciones_lysol_test.values)
print("LÍMITE INFERIOR (90%):")
print(intervalos_lysol.iloc[:, 0].values)
print("LÍMITE SUPERIOR (90%):")
print(intervalos_lysol.iloc[:, 1].values)
```

```python
# Límites de confianza
limite_inferior_vanish = intervalos_vanish.iloc[:, 0]
limite_superior_vanish = intervalos_vanish.iloc[:, 1]

plt.figure(figsize=(12, 5))

# Datos de entrenamiento
plt.plot(train_vanish.index, train_vanish,
         label="Entrenamiento", color="orange")

# Datos reales de prueba
plt.plot(test_vanish.index, test_vanish,
         label="Datos de prueba", color="green")

# Predicciones puntuales
plt.plot(test_vanish.index, predicciones_vanish_test,
         label="Predicciones", color="red")

# Intervalos de confianza
plt.plot(test_vanish.index, limite_inferior_vanish,
         label="Límite inferior", color="blue")

plt.plot(test_vanish.index, limite_superior_vanish,
         label="Límite superior", color="blue")

plt.title(f"Pronóstico de ventas VANISH con ARIMA {mejor_orden_vanish}")
plt.xlabel("Fecha")
plt.ylabel("Unidades vendidas")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Límites de confianza
limite_inferior_lysol = intervalos_lysol.iloc[:, 0]
limite_superior_lysol = intervalos_lysol.iloc[:, 1]

plt.figure(figsize=(12, 5))

# Datos de entrenamiento
plt.plot(train_lysol.index, train_lysol,
         label="Entrenamiento", color="orange")

# Datos reales de prueba
plt.plot(test_lysol.index, test_lysol,
         label="Datos de prueba", color="green")

# Predicciones puntuales
plt.plot(test_lysol.index, predicciones_lysol_test,
         label="Predicciones", color="red")

# Intervalos de confianza
plt.plot(test_lysol.index, limite_inferior_lysol,
         label="Límite inferior", color="blue")

plt.plot(test_lysol.index, limite_superior_lysol,
         label="Límite superior", color="blue")

plt.title(f"Pronóstico de ventas LYSOL con ARIMA {mejor_orden_lysol}")
plt.xlabel("Fecha")
plt.ylabel("Unidades vendidas")
plt.legend()
plt.grid(True)
plt.show()
```
